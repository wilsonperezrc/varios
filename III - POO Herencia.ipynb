##### 1.2. Sobrescritura de Métodos (Override)

La subclase puede redefinir un método heredado. Donde el método definido en la clase hija, sobrescribe al método de la clase padre.

```python
class Animal:
    def sonido(self):
        print("Sonido genérico")


class Perro(Animal):
    def sonido(self):
        print("Guau Guau")
```

### 2. Tipos de Herencia

#### 2.1. Herencia Simple
Una clase hija hereda de una sola clase padre.

```python
class Animal:
    pass

class Perro(Animal):
    pass
```

#### 2.2. Herencia Multinivel
Una clase hereda de otra clase derivada.

```python
class Animal:
    pass

class Mamifero(Animal):
    pass

class Perro(Mamifero):
    pass
```

#### 2.3. Herencia Múltiple
Una clase hereda de varias clases.

```python
class Volador:

    def volar(self):
        print("Volando")

class Nadador:

    def nadar(self):
        print("Nadando")

class Pato(Volador,Nadador):
    pass
```

#### 3.1. Gestión de Vehículos

Una empresa de transporte necesita desarrollar un sistema para administrar diferentes tipos de vehículos.

Todos los vehículos poseen información común como la marca, el modelo y la velocidad actual. Sin embargo, cada tipo de vehículo tiene características particulares.

Se requiere crear una jerarquía de clases que permita representar automóviles y motocicletas, aprovechando la reutilización de código mediante herencia.

- Diseñe una clase que represente a un vehículo.
- Cree clases derivadas para automóvil y motocicleta.
- Permita aumentar y disminuir la velocidad de los vehículos.
- Muestre toda la información de cada objeto creado.
- Cree al menos dos vehículos diferentes para probar el sistema.

```python
                          +----------------------+
                          |       Vehiculo       |
                          +----------------------+
                          | - marca             |
                          | - modelo            |
                          | - velocidad         |
                          +----------------------+
                          | + acelerar()        |
                          | + frenar()          |
                          | + mostrarDatos()    |
                          +----------------------+
                                     ▲
                                     |
                     ----------------------------------
                     |                                |
                     |                                |
          +---------------------+       +----------------------+
          |      Automovil      |       |     Motocicleta      |
          +---------------------+       +----------------------+
          | - puertas           |       | - cilindrada         |
          +---------------------+       +----------------------+
          | + mostrarDatos()    |       | + mostrarDatos()     |
          +---------------------+       +----------------------+
```

```python
# ==========================================
# Clase Padre
# ==========================================

class Vehiculo:

    def __init__(self, marca, modelo, velocidad=0):
        self.marca = marca
        self.modelo = modelo
        self.velocidad = velocidad

    def acelerar(self, cantidad):
        self.velocidad += cantidad
        print(f"El vehículo aceleró {cantidad} km/h.")

    def frenar(self, cantidad):
        self.velocidad -= cantidad

        if self.velocidad < 0:
            self.velocidad = 0

        print(f"El vehículo frenó {cantidad} km/h.")

    def mostrar_datos(self):
        print(f"Marca: {self.marca}")
        print(f"Modelo: {self.modelo}")
        print(f"Velocidad: {self.velocidad} km/h")


# ==========================================
# Clase Hija: Automóvil
# ==========================================

class Automovil(Vehiculo):

    def __init__(self, marca, modelo, velocidad, puertas):
        super().__init__(marca, modelo, velocidad)
        self.puertas = puertas

    def mostrar_datos(self):
        print("\n===== AUTOMÓVIL =====")
        super().mostrar_datos()
        print(f"Puertas: {self.puertas}")


# ==========================================
# Clase Hija: Motocicleta
# ==========================================

class Motocicleta(Vehiculo):

    def __init__(self, marca, modelo, velocidad, cilindrada):
        super().__init__(marca, modelo, velocidad)
        self.cilindrada = cilindrada

    def mostrar_datos(self):
        print("\n===== MOTOCICLETA =====")
        super().mostrar_datos()
        print(f"Cilindrada: {self.cilindrada} cc")


# ==========================================
# Programa Principal
# ==========================================

auto1 = Automovil(
    "Toyota",
    "Corolla",
    0,
    4
)

moto1 = Motocicleta(
    "Yamaha",
    "MT-03",
    0,
    321
)

# Operaciones con el automóvil
auto1.acelerar(80)
auto1.frenar(20)

# Operaciones con la motocicleta
moto1.acelerar(100)
moto1.frenar(30)

# Mostrar resultados
auto1.mostrar_datos()
moto1.mostrar_datos()
```

#### 3.2. Personal de una Empresa

Una empresa tecnológica necesita administrar la información de sus empleados.

Todos los empleados tienen nombre y salario, pero algunos ocupan cargos específicos que requieren información adicional.

Los gerentes administran departamentos y los programadores trabajan con un lenguaje de programación principal.

- Diseñe una clase general para empleados.
- Cree una clase para gerentes.
- Cree una clase para programadores.
- Registre información de varios empleados.
- Muestre los datos completos de cada uno.

```python
                    +----------------------+
                    |      Empleado        |
                    +----------------------+
                    | - nombre             |
                    | - salario            |
                    +----------------------+
                    | + mostrarDatos()     |
                    +----------------------+
                               ▲
                               |
                 ----------------------------
                 |                          |
                 |                          |
      +-------------------+      +----------------------+
      |      Gerente      |      |     Programador      |
      +-------------------+      +----------------------+
      | - departamento    |      | - lenguajeFavorito  |
      +-------------------+      +----------------------+
      | + mostrarDatos()  |      | + mostrarDatos()     |
      +-------------------+      +----------------------+
```

```python
# ==========================================
# Clase Padre
# ==========================================

class Empleado:

    def __init__(self, nombre, salario):
        self.nombre = nombre
        self.salario = salario

    def mostrar_datos(self):
        print(f"Nombre: {self.nombre}")
        print(f"Salario: ${self.salario:.2f}")


# ==========================================
# Clase Hija: Gerente
# ==========================================

class Gerente(Empleado):

    def __init__(self, nombre, salario, departamento):
        super().__init__(nombre, salario)
        self.departamento = departamento

    def mostrar_datos(self):
        print("\n===== GERENTE =====")
        super().mostrar_datos()
        print(f"Departamento: {self.departamento}")


# ==========================================
# Clase Hija: Programador
# ==========================================

class Programador(Empleado):

    def __init__(self, nombre, salario, lenguaje_favorito):
        super().__init__(nombre, salario)
        self.lenguaje_favorito = lenguaje_favorito

    def mostrar_datos(self):
        print("\n===== PROGRAMADOR =====")
        super().mostrar_datos()
        print(f"Lenguaje Favorito: {self.lenguaje_favorito}")


# ==========================================
# Programa Principal
# ==========================================

gerente1 = Gerente(
    "Pedro Gómez",
    2500,
    "Desarrollo de Software"
)

programador1 = Programador(
    "Ana Torres",
    1800,
    "Python"
)

programador2 = Programador(
    "Carlos Pérez",
    1700,
    "Java"
)

# Mostrar información

gerente1.mostrar_datos()

programador1.mostrar_datos()

programador2.mostrar_datos()
```

#### 3.3. Sistema Universitario

Una universidad desea implementar un sistema para gestionar a todos los miembros de su comunidad académica.

Todas las personas tienen nombre y edad. Existen estudiantes y empleados. Los estudiantes pueden pertenecer a programas de pregrado (tiene semestre) o posgrado (nombre de la maestría), mientras que los empleados pueden ser docentes (especialidad) o administrativos (departamento). Un estudiante tiene una carrera y un empleado un salario.

El sistema debe aprovechar la herencia para evitar la duplicación de código y representar correctamente las relaciones entre las distintas entidades.

- Diseñe una jerarquía de clases completa para representar el sistema.
- Utilice herencia multinivel para modelar las relaciones entre las clases.
- Cree al menos un objeto de cada clase.
- Muestre toda la información almacenada.
- Explique qué atributos y métodos son heredados en cada nivel de la jerarquía.
- Elabore un diagrama de clases (UML) que represente la solución propuesta.
- Justifique por qué la herencia es una mejor alternativa que repetir código en este caso.

```python
                           +------------------+
                           |     Persona      |
                           +------------------+
                           | - nombre         |
                           | - edad           |
                           +------------------+
                           | + mostrarDatos() |
                           +------------------+
                                    ▲
                                    |
                  -------------------------------------
                  |                                   |
                  |                                   |
          +---------------+                 +---------------+
          |  Estudiante   |                 |   Empleado    |
          +---------------+                 +---------------+
          | - carrera     |                 | - salario     |
          +---------------+                 +---------------+
          | + mostrar()   |                 | + mostrar()   |
          +---------------+                 +---------------+
                  ▲                                   ▲
                  |                                   |
          -----------------                 -------------------
          |               |                 |                 |
          |               |                 |                 |
 +----------------+  +----------------+ +-------------+ +----------------+
 |   Pregrado     |  |   Posgrado     | |   Docente   | | Administrativo |
 +----------------+  +----------------+ +-------------+ +----------------+
 | - semestre     |  | - maestria     | | - materia   | | - departamento |
 +----------------+  +----------------+ +-------------+ +----------------+
 | + mostrar()    |  | + mostrar()    | | + mostrar() | | + mostrar()    |
 +----------------+  +----------------+ +-------------+ +----------------+
```

```python
# Solución

# ==========================================
# Clase Padre
# ==========================================

class Persona:

    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def mostrar_datos(self):
        print(f"Nombre: {self.nombre}")
        print(f"Edad: {self.edad}")


# ==========================================
# Nivel 2
# ==========================================

class Estudiante(Persona):

    def __init__(self, nombre, edad, carrera):
        super().__init__(nombre, edad)
        self.carrera = carrera

    def mostrar_datos(self):
        super().mostrar_datos()
        print(f"Carrera: {self.carrera}")


class Empleado(Persona):

    def __init__(self, nombre, edad, salario):
        super().__init__(nombre, edad)
        self.salario = salario

    def mostrar_datos(self):
        super().mostrar_datos()
        print(f"Salario: ${self.salario:.2f}")


# ==========================================
# Nivel 3 (Herencia Multinivel)
# ==========================================

class Pregrado(Estudiante):

    def __init__(self, nombre, edad, carrera, semestre):
        super().__init__(nombre, edad, carrera)
        self.semestre = semestre

    def mostrar_datos(self):
        print("\n===== ESTUDIANTE DE PREGRADO =====")
        super().mostrar_datos()
        print(f"Semestre: {self.semestre}")


class Posgrado(Estudiante):

    def __init__(self, nombre, edad, carrera, maestria):
        super().__init__(nombre, edad, carrera)
        self.maestria = maestria

    def mostrar_datos(self):
        print("\n===== ESTUDIANTE DE POSGRADO =====")
        super().mostrar_datos()
        print(f"Maestría: {self.maestria}")


class Docente(Empleado):

    def __init__(self, nombre, edad, salario, materia):
        super().__init__(nombre, edad, salario)
        self.materia = materia

    def mostrar_datos(self):
        print("\n===== DOCENTE =====")
        super().mostrar_datos()
        print(f"Materia: {self.materia}")


class Administrativo(Empleado):

    def __init__(self, nombre, edad, salario, departamento):
        super().__init__(nombre, edad, salario)
        self.departamento = departamento

    def mostrar_datos(self):
        print("\n===== ADMINISTRATIVO =====")
        super().mostrar_datos()
        print(f"Departamento: {self.departamento}")


# ==========================================
# Programa Principal
# ==========================================

est_pregrado = Pregrado(
    "Juan Pérez",
    20,
    "Ciencia de Datos e Inteligencia Artificial",
    3
)

est_posgrado = Posgrado(
    "María López",
    28,
    "Ingeniería en Software",
    "Maestría en Inteligencia Artificial"
)

docente = Docente(
    "Pedro García",
    45,
    1800,
    "Programación Orientada a Objetos"
)

administrativo = Administrativo(
    "Ana Torres",
    38,
    1200,
    "Talento Humano"
)

# Mostrar información

est_pregrado.mostrar_datos()

est_posgrado.mostrar_datos()

docente.mostrar_datos()

administrativo.mostrar_datos()
```

#### 3.4. Sistema de Vehículos Inteligentes

Una empresa automotriz está desarrollando una nueva generación de vehículos inteligentes que integran diferentes tecnologías para mejorar la experiencia de conducción.

Dentro de sus nuevos modelos existen vehículos que poseen características especiales. Algunos funcionan mediante energía eléctrica, otros son capaces de conducirse de forma autónoma y ciertos modelos combinan ambas funcionalidades.

Se requiere desarrollar una aplicación orientada a objetos que permita representar un vehículo inteligente capaz de heredar funcionalidades de varias clases al mismo tiempo.

Requerimientos
- El tipo Electrico que almacene el porcentaje de batería disponible y permita cargar la batería del vehículo.
- El tipo Autonomo que almacene el nivel de autonomía del vehículo y permita activar el modo de conducción autónoma.
- El nuevo tipo de auto deberá almacenar además la marca y modelo.
- Implemente un método que muestre toda la información del vehículo.
- Cree un objeto de tipo AutoInteligente y pruebe todas sus funcionalidades heredadas.

Actividades
- Crear las clases necesarias.
- Implementar la herencia múltiple.
- Crear un automóvil inteligente.
- Cargar la batería.
- Activar la conducción autónoma.
- Mostrar toda la información del vehículo.
- Identificar qué atributos y métodos provienen de cada clase padre.



```python
                    +----------------------+
                    |      Electrico       |
                    +----------------------+
                    | - bateria            |
                    +----------------------+
                    | + cargar()           |
                    +----------------------+
                              ▲
                              |
                              |
                              |
                              |
                    +----------------------+
                    |   AutoInteligente    |
                    +----------------------+
                    | - marca              |
                    | - modelo             |
                    +----------------------+
                    | + mostrarDatos()     |
                    +----------------------+
                              |
                              |
                              |
                              ▼
                    +----------------------+
                    |      Autonomo        |
                    +----------------------+
                    | - nivelAutonomia     |
                    +----------------------+
                    | + conducir()         |
                    +----------------------+
```

```python
# ==========================================
# Clase Padre 1
# ==========================================

class Electrico:

    def __init__(self, bateria):
        self.bateria = bateria

    def cargar(self):
        self.bateria = 100
        print("Batería cargada al 100%.")


# ==========================================
# Clase Padre 2
# ==========================================

class Autonomo:

    def __init__(self, nivel_autonomia):
        self.nivel_autonomia = nivel_autonomia

    def conducir(self):
        print(f"Conduciendo en modo autónomo nivel {self.nivel_autonomia}.")


# ==========================================
# Clase Hija (Herencia Múltiple)
# ==========================================

class AutoInteligente(Electrico, Autonomo):

    def __init__(self, marca, modelo, bateria, nivel_autonomia):

        Electrico.__init__(self, bateria)
        Autonomo.__init__(self, nivel_autonomia)

        self.marca = marca
        self.modelo = modelo

    def mostrar_datos(self):
        print("\n===== AUTO INTELIGENTE =====")
        print(f"Marca: {self.marca}")
        print(f"Modelo: {self.modelo}")
        print(f"Batería: {self.bateria}%")
        print(f"Nivel de autonomía: {self.nivel_autonomia}")


# ==========================================
# Programa Principal
# ==========================================

auto1 = AutoInteligente(
    "Tesla",
    "Model X",
    80,
    5
)

auto1.mostrar_datos()

print("\n--- Funcionalidades heredadas ---")
auto1.cargar()
auto1.conducir()

print("\n--- Estado final ---")
auto1.mostrar_datos()
```
