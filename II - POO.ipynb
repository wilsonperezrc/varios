#### 1. Clases en Python


##### 1.1. Definir Clases
Para definir una clase en Python lo haremos de la siguiente forma:




```python
class Saludo:
    pass
```

***Constructores:*** Un constructor es un método o función especial dentro de una clase que se ejecuta automáticamente al crear (instanciar) un nuevo objeto

Se define con \_\_init\_\_ (self)

```python
class Saludo:

    def __init__(self):
        print('Me acaban de instanciar desde la clase Saludo')
```

Su función principal es asignar los valores iniciales a los atributos del objeto y reservar los recursos necesarios para su uso.

***self:*** Este parámetro, también conocido como this en otros lenguajes, se debe pasar obligatoriamente como primer parámetro en TODOS los métodos que creemos dentro de la clase y lo utilizaremos para acceder a todos los atributos y métodos de la clase.

##### 1.4. Métodos

Los métodos son las funciones que definen el comportamiento del objeto. Representan lo que el objeto hace.

```python
class Saludo:

    nombre: str
    def __init__(self, nombre):
        self.nombre = nombre
        print('Asigno al atributo nombre, el valor de ', nombre)

    def saludar(self):
        print(f'Hola {self.nombre}')
```

```python
saludo = Saludo('Alber')
saludo.saludar()
```

```python
saludo.nombre
```

```python
saludo = Saludo('Lucas')
saludo.saludar()
```

#### Ejemplo 7: Ejemplo completo I

```python
from datetime import datetime

class Persona:

    def __init__(self):
        self.nombres = ""
        self.apellidos = ""
        self.ciudad = ""
        self.direccion = ""
        self.fecha_nacimiento = ""
        self.tipo_sangre = ""

    def ingresar_datos(self, nombres, apellidos, ciudad, direccion, fecha_nacimiento, tipo_sangre):

        self.nombres = nombres
        self.apellidos = apellidos
        self.ciudad = ciudad
        self.direccion = direccion
        self.fecha_nacimiento = fecha_nacimiento
        self.tipo_sangre = tipo_sangre

    def mostrar_datos(self):
        print("\n=== DATOS DE LA PERSONA ===")
        print("Nombres:", self.nombres)
        print("Apellidos:", self.apellidos)
        print("Ciudad:", self.ciudad)
        print("Dirección:", self.direccion)
        print("Fecha de nacimiento:", self.fecha_nacimiento)
        print("Tipo de sangre:", self.tipo_sangre)
        print("Edad:", self.calcular_edad(), "años")

    def calcular_edad(self):
        try:
            fecha_nac = datetime.strptime(
                self.fecha_nacimiento, "%d/%m/%Y")
            hoy = datetime.today()
            edad = hoy.year - fecha_nac.year

            if (hoy.month, hoy.day) < (fecha_nac.month, fecha_nac.day):
                edad -= 1
            return edad

        except:
            return "Fecha inválida"

    def actualizar_direccion(self, nueva_direccion):
        self.direccion = nueva_direccion

    def actualizar_ciudad(self, nueva_ciudad):
        self.ciudad = nueva_ciudad
```

```python
persona1 = Persona()
persona1.ingresar_datos('Juan Esteban', 'Cardenas Encalada', 'Ibarra', 'Av 12 Junio', '20/01/1995', 'O+')
```

```python
persona1.mostrar_datos()
```

```python
persona1.calcular_edad()
```

```python
persona1.actualizar_direccion("Calle Quito y Ambato")
```

```python
persona1.mostrar_datos()
```

```python
persona1.actualizar_ciudad("Quito")
```

```python
persona1.mostrar_datos()
```
