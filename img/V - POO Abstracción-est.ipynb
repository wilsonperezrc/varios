#### 1. Abstracción en Python

```python
from abc import ABC, abstractmethod

class Figura(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura
```

```python
r = Rectangulo(5, 4)
print(r.area())
```

### Ejemplos

#### Ejemplo 1:

```python
from abc import ABC, abstractmethod

class Vehiculo(ABC):
    @abstractmethod
    def mover(self):
        pass

class Auto(Vehiculo):
    def mover(self):
        print("El auto se desplaza por la carretera")

class Avion(Vehiculo):
    def mover(self):
        print("El avión vuela por el aire")
```

```python
a = Auto()
v = Avion()

a.mover()
v.mover()
```

#### Ejemplo 2:

```python
from abc import ABC, abstractmethod

class Figura(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimetro(self):
        pass

class Rectangulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return (self.base + self.altura) * 2

class TrianguloRectandulo(Figura):
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return (self.base * self.altura) / 2

    def perimetro(self):
        h =(self.base** 2 + self.altura**2) ** (1/2)
        return self.base + self.altura + h
```

```python
r = Rectangulo(5, 4)
print("El área del cuadrado es : ", r.area())
print("El perímetro del cuadrado es : ", r.perimetro())

t = TrianguloRectandulo(5, 4)
print("El área del triángulo es : ", t.area())
print("El perímetro del triángulo es : ", t.perimetro())
```

#### Ejemplo 3:

```python
class Figura(ABC):
    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass

class Rectangulo(Figura):

    def __init__(self, color, base, altura):
        super().__init__(color)
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura

r = Rectangulo("Rojo", 5, 4)

print("Color:", r.color)
print("Base:", r.base)
print("Altura:", r.altura)
print("Área:", r.area())
```
