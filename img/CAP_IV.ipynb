#### 4.1.1. TUPLAS

Una tupla es una secuencia inmutable de elementos. Cada uno de los elementos que conforman una tupla puede ser de cualquier tipo (int, string, list, etc.). La sintaxis para declarar una tupla consiste en especificar sus elementos separados por una coma (,).

A menudo se coloca una coma al final de la tupla para indicar la posición del último elemento.


```python

tupla_vacia = ( )
print(tupla_vacia)
```

```python
tupla1 = 1 ,2 ,3 ,4 ,6 ,
print(tupla1)

tupla1b = (1 ,2 ,3 ,4 ,6)
print(tupla1)
```

```python
tupla2 = 'hola' ,2 ,3
print(tupla2)
```

```python
tupla3 = tupla2 ,90
print(tupla3)
```

```python
tupla4 = tuple ( 'ejemplo ' ) ;
print(tupla4)
```

#### 4.1.2. LISTAS

Una lista es una secuencia mutable de elementos. A diferencia de las tuplas, los elementos de las listas se pueden modificar una vez han recibido un valor asignado.

- Para especificar los elementos que forman una lista, se usa una tira de elementos separada de nuevo por coma (,), al principio y al final la lista se envuelve entre corchetes [,].


```python
lista_vacia = [ ] ;
print(lista_vacia)
```

```python
lista = [1 ,2 ,3 ,4 ,6]
print(lista)
```

- Para añadir elementos a una lista, se dispone de dos métodos, append y extend. Append añade un elemento a una lista (independientemente de su tipo), y extend recibe una lista de elementos, y los concatena a la lista actual.

```python
lista = [1 ,2 ,3 ,4 ,6]
print(lista)

lista.append ( [ 7 , 8 , 9 ] )
print(lista)
```

```python
lista = [1 ,2 ,3 ,4 ,6]
lista.extend ( [ 7 , 8 , 9 ] )
print(lista)
```

```python
## Ejemplo 1: Ingresar 5 números y agregar a una lista.

datos = []

for i in range(3):
    valor = int(input("Ingrese un número: "))
    datos.append(valor)

print(datos)

datos.extend([100, 200])

print(datos)
```

```python
## Ejemplo 2: En un rango de q al 10, agregar números impares con append y los números pares con extend.
numeros_pares = []
numeros_impares = []

for i in range(1, 11):
    if i % 2 != 0:
        numeros_impares.append(i) # Agregar números impares con append()
    else:
        numeros_pares.extend([i]) # Agregar números pares con extend()

print(numeros_impares)
print(numeros_pares)
```

```python
## Ejemplo 3: Crea dos listas vacías, una con números positivos y otra con números negativos. Pide al usuario ingresar números hasta que ingrese 0 y usa append() para guardarlos según corresponda.

positivos = []
negativos = []

while True:
    numero = int(input("Ingrese un número, 0 para terminar: "))

    if numero == 0:
        break
    elif numero > 0:
        positivos.append(numero)
    else:
        negativos.append(numero)

print("Números positivos:", positivos)
print("Números negativos:", negativos)
```

#### 4.1.3. DICCIONARIOS

Los diccionarios conforman uno de los tipos de datos más útiles para los programadores en Python. Se trata de unas estructuras de memoria asociativa. Los elementos en los diccionarios no se consultan mediante un índice, sino que se accede directamente a ellos por contenido.

Para crear el diccionario simplemente indicaremos entre llaves {,} las parejas de elementos que lo conforman. En cada pareja indicaremos primero el valor 'llave' para acceder al elemento, y después el valor que contendrá (que puede
ser de cualquiera de los tipos disponibles en Python).

- Clave → identificador (único)
- Valor → dato asociado a la clave

```python
diccionario1 = { 'nombre' : 'Alejandro' ,
                 'edad' :   34 ,
                 'numerosfavoritos' : (3 ,7 ,13) }
print(diccionario1)
```

```python
diccionario0 = { 'nombre' : 'Alejandro' , 'nombre' : 'Juanita' ,
                 'edad' :   34 ,
                 'numerosfavoritos' : (3 ,7 ,13) }
print(diccionario0)
```

#### 4.1.3.1. Ejemplos

```python
## Ejemplo 1: Diccionario de notas

notas = {
    "Matemáticas": 8,
    "Programación": 9,
    "Física": 7
}

suma = 0
for nota in notas.values():
    suma += nota

promedio = suma / len(notas)
print("Promedio:", promedio)
```

```python
## Ejemplo 2: Contar ocurrencias de las letras

frase = "hola mundo"
contador = {}

for letra in frase:
    if letra in contador:
        contador[letra] += 1
    else:
        contador[letra] = 1

print(contador)
```

```python
## Ejemplo 3: Lista en diccionario
estudiante = {
    "nombre": "Ana",
    "notas": [8, 9, 7]
}

suma = 0
for nota in estudiante["notas"]:
    suma += nota

promedio = suma / len(estudiante["notas"])
print("Promedio de calificaciones:", promedio)
```
