### 3.1. Visualización de datos

#### ¿Qué es la visualización de datos?

La visualización de datos es el proceso de representar información de forma gráfica para:

- Facilitar la interpretación
- Detectar patrones
- Identificar anomalías
- Apoyar la toma de decisiones

#### Tipos de gráficos

##### Gráfico de líneas: Representa evolución en el tiempo

- Series temporales
- Tendencias

#### Gráfico de barras: Comparación entre categorías

- Conteos
- Frecuencias
- Histogramas

#### Scatter plot: Relación entre variables

- Correlación
- Patrones

#### Boxplot: Resumen estadístico

- Mediana
- Cuartiles
- Outliers


### 3.2. Principios del Diseño Visual

- Claridad: Evitar ruido visual
- Simplicidad: Menos es más
- Consistencia: Colores y estilos coherentes
- Precisión: No distorsionar datos
- Jerarquía visual: Resaltar lo importante


### 3.3. Matplotlib

El paquete matplotlib es una biblioteca de funciones para generar gráficos 2D y 3D.

Características son:

- Curva de aprendizaje suave: ofrece funcionalidades simples para usuarios ocasionales.

- Todo tipo de ajustes programables de los elementos de una ventana gráfica.

- Gran variedad de formatos de exportación de las figuras.

Matplotlib está construido sobre el paquete Numpy y trabaja de forma natural con los vectores y matrices (arrays) a ella asociados.

### Propiedades de plot()
La configuración se realiza con funciones

#### Propiedades de **plt.plot()**
- Título: **plt.title("Título")**
- Etiqueta eje X: **plt.xlabel("Mes")**
    - **plt.xlabel("Mes", fontsize=12)**
- Etiqueta eje Y: **plt.ylabel("Ventas")**
- Leyenda: **plt.legend()**
- Límites en x: **plt.xlim(0, 10)**
- Límites en y: **plt.ylim(0, 20)**
- Cuadrícula: **plt.grid(True)**
- Tamaño: **plt.figure(figsize=(8,5))**
- Guardar: **plt.savefig("Figure.pdf")**

#### Propiedades dentro de **plt.plot()**


```python
plt.plot(
    x, y,
    color="red",       # color línea
    linestyle="--",    # línea punteada
    linewidth=2,       # grosor
    marker="o",        # círculo en puntos
    markersize=8,      # tamaño círculo
    label="Datos"      # nombre leyenda
)
```

- linestyle:

| Valor  | Significado |

- marker:
| Valor | Forma     |
| ----- | --------- |
| `"o"` | círculo   |
| `"s"` | cuadrado  |
| `"^"` | triángulo |
| `"x"` | cruz      |
| `"*"` | estrella  |
| `"."` | punto     |

```python
def plot_vectors3d(ax, vectors3d, z0, **options):
    for v in vectors3d:
        x, y, z = v
        ax.plot([x,x], [y,y], [z0, z], color="gray", linestyle='dotted', marker=".")
    x_coords, y_coords, z_coords = zip(*vectors3d)
    ax.scatter(x_coords, y_coords, z_coords, **options)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

subplot3d = plt.subplot(111, projection='3d')
subplot3d.set_zlim([0, 9])
plot_vectors3d(subplot3d, [a,b], 0, color=("r","b"))
plt.show()
```

### *Scatter*
La función `scatter` nos permite visualizar un conjunto de puntos.

```python
from numpy.random import rand
x, y = rand(2, 100)
plt.scatter(x, y)
plt.show()
```

```python
for color in ['red', 'green', 'blue']:
    n = 100
    x, y = rand(2, n)
    scale = 500.0 * rand(n) ** 5
    plt.scatter(x, y, s=scale, c=color, alpha=0.3)

plt.grid(True)

plt.show()
```
