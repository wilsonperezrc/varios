### 1. Ciclo While

#### Ejemplo 1: Imprimir los 9 primeros números en un ciclo “Mientras”.

```python
# Inicializar variable
i = 1

# Bucle mientras
while i <= 9:
    print(i)
    i = i + 1
```

#### Ejemplo 2: Sumar los 9 primeros números en un ciclo “Mientras”.

```python
# Inicializar variables
suma = 0
i = 1

# Bucle mientras
while i <= 9:
    suma = suma + i
    i = i + 1

# Mostrar resultado
print("La suma es:", suma)
```

#### Ejemplo 3: Obtener la suma de diez cantidades mediante la utilización de un ciclo “Mientras”.

```python
# Inicializar variables
suma = 0
i = 1

# Ciclo para leer 10 números y sumarlos
while i <= 10:
    n = float(input("Ingrese un número: "))
    suma = suma + n
    i = i + 1

# Mostrar resultado
print("La suma es:", suma)
```

### 2. Ciclo Do - While

#### Ejemplo 5: Imprimir los 9 primeros números en un ciclo “Hacer - Mientras”.

```python
# Inicializar variable
i = 1

# Simulación de Hacer-Mientras
while True:
    print(i)
    i = i + 1

    if i > 9:
        break
```

```python
# Tarea 02.- Imprimir los números en orden ascendente y descendente en un rango entre inicio y fin.
```

```python
print('Ingrese lim inferior: ')
n1 = int(input())

print('Ingrese lim superior: ')
n2 = int(input())

print('Lim inferior:', n1)
print('Lim superior:', n2)

print('Orden descendente: ')
i = n2

while i >= n1 :
    print(i)
    i=i-1

print('Orden ascendente: ')
j = n1

while j <= n2 :
    print(j)
    j = j + 1
```

```python
# Tarea 03.- Realizar una aplicación que calcule el Factorial de un número
```

```python
print('Ingrese número: ')
n = int(input())
print(n)

fac = 1

while n >= 1:
    fac = fac * n
    n = n - 1

print('El factorial es', fac)
```

```python
# Tarea 04.- Realizar una aplicación que calcule la Potencia de un número base elevado a un exponente 
```

```python
print('Ingrese el número base: ')
base = int(input())
print(base)

print('Ingrese el exponente: ')
ex = int(input())
print(ex)

res = 1

while ex >= 1:
    res = res * base
    ex = ex - 1

print('El exponente es:', res)
```

```python
# Tarea 05.- Imprimir todas las potencias del número num, elevado a los números de un rango definido (ini-fin) 
```

```python
print('Ingrese el número base: ')
num = int(input())
print(num)

print('Ingrese el rango inicial ')
ini = int(input())
print(ini)

print('Ingrese el rango final ')
fin = int(input())
print(fin)

print('Salida ')
while fin >= ini :
    res = num ** ini
    print(num, 'elevado a', ini, 'es',res)
    ini = ini + 1
```

```python
# Ej 6: Imprimir los números pares e impares en un rango
```

```python
print('Ingrese el número inferior: ')
inf = int(input())
print(inf)

print('Ingrese el rango superior ')
sup = int(input())
print(sup)

i = inf

while i <= sup :
    res = i % 2

    if res == 0:
        print(i, 'Es par')
    else :
        print(i, 'Es impar')

    i = i + 1
```
