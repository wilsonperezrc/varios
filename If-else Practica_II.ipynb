### 2.1. If - elif - else
***

Estas son estructuras de decisión en Python que permiten que un programa elija qué código ejecutar dependiendo de una condición.

Sirven para que un programa tome decisiones lógicas:

- if → pregunta inicial (¿se cumple la condición?)

- elif → si el primer if no se cumple, prueba otra condición

- else → si ninguna condición se cumple, ejecuta otra acción

```python
if condicion1:
    # código si condicion1 es verdadera
elif condicion2:
    # código si condicion1 no se cumplió, pero condicion2 sí
elif condicion3:
    # más opciones...
else:
    # código si ninguna condición se cumple
```

### 2.2. Match (switch)
***

En Python no existe la estructura tradicional switch como en otros lenguajes (Java, C, C++).

Sin embargo, sí existen alternativas match



```python
match variable:
    case valor1:
        ...
    case valor2:
        ...
    case _:
        ...
```

#### EJEMPLO

```python
num = 7

match num:
    case 1:
        print("Lunes")
    case 2:
        print("Martes")
    case 3:
        print("Miércoles")
    case 4:
        print("Jueves")
    case 5:
        print("Viernes")
    case 6:
        print("Sábado")
    case 7:
        print("Domingo")
    case _:
        print("Número fuera de rango")
```
