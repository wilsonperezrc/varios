***Constructores:*** Un constructor es un método o función especial dentro de una clase que se ejecuta automáticamente al crear (instanciar) un nuevo objeto

Se define con \_\_init\_\_ (self)

```python
class Saludo:

    def __init__(self):
        print('Me acaban de instanciar desde la clase Saludo')
```

#### Ejemplo 7: Ejemplo completo I

```python
from datetime import datetime

class Persona:

    def __init__(self):
        self.nombres = ""
        self.apellidos = ""
        self.ciudad = ""
        self.direccion = ""
        self.fecha_nacimiento = ""
        self.tipo_sangre = ""

    def ingresar_datos(self, nombres, apellidos, ciudad, direccion, fecha_nacimiento, tipo_sangre):

        self.nombres = nombres
        self.apellidos = apellidos
        self.ciudad = ciudad
        self.direccion = direccion
        self.fecha_nacimiento = fecha_nacimiento
        self.tipo_sangre = tipo_sangre

    def mostrar_datos(self):
        print("\n=== DATOS DE LA PERSONA ===")
        print("Nombres:", self.nombres)
        print("Apellidos:", self.apellidos)
        print("Ciudad:", self.ciudad)
        print("Dirección:", self.direccion)
        print("Fecha de nacimiento:", self.fecha_nacimiento)
        print("Tipo de sangre:", self.tipo_sangre)
        print("Edad:", self.calcular_edad(), "años")

    def calcular_edad(self):
        try:
            fecha_nac = datetime.strptime(
                self.fecha_nacimiento, "%d/%m/%Y")
            hoy = datetime.today()
            edad = hoy.year - fecha_nac.year

            if (hoy.month, hoy.day) < (fecha_nac.month, fecha_nac.day):
                edad -= 1
            return edad

        except:
            return "Fecha inválida"

    def actualizar_direccion(self, nueva_direccion):
        self.direccion = nueva_direccion

    def actualizar_ciudad(self, nueva_ciudad):
        self.ciudad = nueva_ciudad
```

#### Ejemplo 8: Mantenimiento de productos

Un producto posee nombre, precio y stock.  A cada producto se puede vender, consultar precio, consultar stock e incrementar su número de productos a stock

```python
class Producto:

    def __init__(self, nombre, precio, stock):
        self.nombre = nombre
        self.precio = precio
        self.stock = stock

    def vender(self, cantidad):
        if cantidad <= self.stock:
            self.stock -= cantidad
            print(f"Se vendieron {cantidad} unidades.")
        else:
            print("Stock insuficiente.")

    def abastecer(self, cantidad):
        self.stock = self.stock + cantidad
        print(f"Se agregaron {cantidad} unidades al inventario.")

    def consultarNombre(self):
        return(self.nombre)

    def consultarStock(self):
        return(self.stock)

    def consultarPrecio(self):
        return(self.precio)
```

```python
tv = Producto('TV', '350', 100)
tv_numProductos = tv.consultarPrecio()
print(f'El número total de {tv.consultarNombre()} en stock es {tv_numProductos}')
tv_precio = tv.consultarStock()
print(f'Existen {tv_numProductos} del producto {tv.consultarNombre()} en stock')
```

```python
tv.abastecer(50)
print(f'Existen {tv.consultarStock()} del producto {tv.consultarNombre()} en stock')
```

```python
tv.vender(500)
```

```python
tv.vender(30)
```

```python
print(f'Existen {tv.consultarStock()} del producto {tv.consultarNombre()} en stock')
```

#### Ejemplo 9: Mantenimiento de Puntos de vida en videojuegos

```python


class Jugador:

    def __init__(self, nombre):
        self.nombre = nombre
        self.nivel = 1
        self.experiencia = 0

    def ganarExperiencia(self, puntos):
        self.experiencia += puntos
        print(f"{self.nombre} ganó {puntos} puntos de experiencia.")

    def subirNivel(self):
        # Regla: cada 100 puntos de experiencia sube un nivel
        while self.experiencia >= 100:
            self.nivel += 1
            self.experiencia = 0

        print(f"{self.nombre} subió de nivel.")

    def imprimirDatos(self):
        print("\n=== DATOS DEL JUGADOR ===")
        print("Nombre:", self.nombre)
        print("Nivel:", self.nivel)
        print("Experiencia:", self.experiencia)

    def consultarNombre(self):
        return (self.nombre)

    def consultarNivel(self):
        return (self.nivel)

    def consultarExperiencia(self):
        return (self.experiencia)
```

```python
jugador1 = Jugador('jugador1')
print (f'El jugador {jugador1.consultarNombre()} tiene el nivel {jugador1.consultarNivel()}, con la experiencia {jugador1.consultarExperiencia()}')
```
